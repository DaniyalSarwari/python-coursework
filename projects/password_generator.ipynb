## Password Generator

Password Generator can automatically create strong, random passwords. Users should be able to specify the **length** of the password and the **number** of passwords they want to generate, ensuring better security for their accounts.

Key Objectives:

- Generate passwords of user-defined length.
- Generate a user-defined number of passwords.
- Ensure randomness to enhance password strength.
- Provide a simple interface for easy use.

## Solution

```python
import random

def generate_character() -> str:
  """
  This function will generate a character based on ASCII value
  """
  ascii_value: int = random.randint(33, 126)
  character: str = chr(ascii_value)

  return character

def generate_password(l: int) -> str:
  password: str = ''

  for _ in range(l):
    password += generate_character()

  return password

def get_valid_number(text: str) -> int:

  while True:
    try:
      numbr: int = int(input(text))
      if isinstance(numbr, int) and numbr > 0:
        break
      else:
        raise Exception("NONACCEPTABLE")

    except Exception as e:
      if str(e) == "NONACCEPTABLE":
        print("Please Enter a Number greater then 0")
      else:
        print("Please Enter a valid Number")

  return numbr

def main():
  print("===Welcome To Password Generator===")

  passwords: list = []

  amount: int = get_valid_number("Enter amount of passwords to generate? ")
  length: int = get_valid_number("Input you password length: ")

  for _ in range(amount):
    password: str = generate_password(length)
    passwords.append(password)

  if passwords:
    print("\n***Here are your PASSWORDS***")
    for pwd in passwords:
      print(pwd)

if __name__=='__main__':
  main()
```
