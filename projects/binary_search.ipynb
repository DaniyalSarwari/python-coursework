## Binary Search

You are given a sorted array of integers arr in ascending order and an integer target.

Your task is to determine the index of the target element in the array using the Binary Search algorithm.

If the target element is not present in the array, return -1.



## Solution

```python
def binary_search(arr: list, target: int, low=None, high= None) -> int:
  if low is None:
    low = 0
  if high is None:
    high = len(arr) - 1

  middle: int = int((low + high) / 2)
  if target not in arr:
    return -1
  else:
    if arr[middle] == target:
      return middle
    elif target < arr[middle]:
      return binary_search(arr, target=target, low= low, high=middle -1)
    elif target > arr[middle]:
      return binary_search(arr, target=target, low=middle + 1, high=high)

def main():
  sorted_list: list = [2, 4, 5, 6, 7, 9, 11, 15, 20, 22, 24, 25, 26]
  print(f"List is {sorted_list}")
  target: int = int(input("Please Enter a number to search: "))

  index: int = binary_search(sorted_list, target)
  if index == -1:
    print("Given Number is not present in the list")
  else:
    print(f"The required number {target} is at location {index}")

if __name__=="__main__":
  main()
```
