## CountDown Timer

Users need a simple way to track time. This project implements a Countdown Timer that continuously ticks down from a set duration and alerts the user when time is up.

## Solution

```python
import time
from IPython.display import clear_output

def countdown(t: int) -> None:
  clear_output(wait=True)
  while t:
    min_sec: tuple[int,int] = divmod(t, 60)
    min, sec = min_sec
    clock = f"{min:02d}:{sec:02d}"
    print(clock)
    time.sleep(1)
    t -= 1
    clear_output(wait=True)
  print("Time Up!")

if __name__=='__main__':
  # print(help(divmod))
  seconds: int = -1
  while not seconds > 0:
    try:
      seconds: int = int(input("Enter time is seconds: "))
    except Exception as e:
      print("Please Enter a valid Positive Integer")
      seconds = -1

  countdown(seconds)
```
